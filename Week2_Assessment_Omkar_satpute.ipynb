```python
"""Q29. Add a new column funding_efficiency = amount / (2024 - founded_year). Higher = raised more
capital per year of existence. Sort by funding_efficiency descending. Print top 10 with name, industry,
amount, founded_year, and funding_efficiency."""


df["funding_efficiency"] = df["amount"]/ (2024 - df["founded_year"])
sort = df.sort_values("funding_efficiency",ascending=False)

print(sort[[
    "startup_name",
    "industry",
    "amount",
    "founded_year",
    "funding_efficiency"
]].head(10))


```

```python
"""Q30. Using .apply() with a lambda, add a funding_tier column based on amount:
amount ≥ 1000 Cr → 'Unicorn'
amount ≥ 500 Cr → 'Mega'
amount ≥ 100 Cr → 'Large'
amount ≥ 10 Cr → 'Mid'
else (including NaN) → 'Small / Unknown'
Print the count of startups in each tier."""


df["funding_tier"] = df["amount"].apply(
    lambda amount: "Unicorn" if amount >= 1000
    else "Mega" if amount >= 500
    else "Large" if amount >= 100
    else "Mid" if amount >= 10
    else "Small / Unknown"
)

print(df["funding_tier"].value_counts())
```

```python
"""Q31. Using pd.cut(), bin founded_year into 3 eras:
Before 2010 → 'Early'
2010 to 2017 → 'Growth'
2018 and later → 'Recent'
Print the count of startups in each era."""

result= pd.cut(df["founded_year"],bins=[1900,2009,2017,2024],labels = ["Early","Growth","Recent"])

print(result.value_counts())
      

```

```python
"""32. The investors column has multiple investors separated by ', ' (comma + space). Create a new
column investor_count showing the number of investors per startup. Use .apply() with a lambda. Handle
NaN values (count = 0). Print the value_counts() of investor_count."""


df["investor_count"] = df["investors"].apply(
    lambda x: len(x.split(", ")) if pd.notna(x) else 0
)

print(df["investor_count"].value_counts())
```

```python
"""Q33. Compute the average funding amount per industry (ignoring NaN). Sort descending. Print the top 5
industries by average funding."""

avg = df.groupby("industry")["amount"].mean().sort_values(ascending=False)

print(avg.head(5))
```

```python
"""Q34. For each city, compute three metrics in a single groupby using .agg() with a dictionary:
(a) Number of startups
(b) Total funding raised (sum of amount)
(c) Average funding per startup
Print the top 10 cities by number of startups."""


metrics = {
    "startup_name": "count",
    "amount": ["sum", "mean"]
}

output = df.groupby("city").agg(metrics)

output.columns = ["number_of_startups", "total_funding", "average_funding"]

print(output.sort_values("number_of_startups", ascending=False).head(10))
```

```python
"""Q35. Interview-style. Which industry has the highest average funding, but only considering industries
with at least 15 startups? (Sparse industries bias the mean.) Print industry name, count, and average
funding."""

output = df.groupby("industry").agg(count = ("startup_name","count"),
                                   average_funding = ("amount","mean"))
output = output[output["count"] >= 15]

output = output.sort_values("count",ascending=True)

print(output)
```

```python
"""Q36. Using pd.crosstab(), show the distribution of funding_tier across the top 5 industries by count.
Rows = industry, columns = funding_tier, values = count of startups."""

output = pd.crosstab(df["industry"],df["funding_tier"])
E = df["industry"].value_counts().head(5).index

output = output.loc[E]
print(output)
```

```python
"""Q37. For each industry, find the startup with the highest funding amount. Print startup_name, industry,
city, and amount. Hint: groupby + idxmax() on amount."""


index = df.groupby("industry")["amount"].idxmax()

High_funding_amount = df.loc[index,["startup_name", "industry", "city", "amount"]]

print(High_funding_amount)
```

```python
"""Q38. Emerging player finder. An 'emerging player' is a startup founded in 2019 or later, with funding ≥
100 Cr, in one of Fintech / Edtech / HealthTech. Find all of them, sort by amount descending, print
name, industry, city, founded_year, and amount. Why might a VC analyst care about this list?"""



emerging_players = df[(df["founded_year"] >= 2019) & (df["amount"] >= 100) & (df["industry"].isin(["Fintech", "Edtech", "HealthTech"]))]

emerging_players = emerging_players.sort_values("amount",ascending=False)

print(emerging_players[["startup_name", "industry", "city", "founded_year", "amount"]])
```

```python
"""Q39. Investor hub analysis. A foreign investor wants to know which Indian city has the highest TOTAL
funding raised, considering only cities with at least 20 startups (so the number is meaningful). Print city
name, count, and total funding."""

output = df.groupby("city").agg(
    count=("startup_name", "count"),
    total_funding=("amount", "sum")
)

output = output[output["count"] >= 20]

output = output.sort_values("total_funding", ascending=False)

print(output)
```
